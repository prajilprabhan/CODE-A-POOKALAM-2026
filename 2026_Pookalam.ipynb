#Inner Pookalam Design


```python
p1 = point(x=10, y=25)
p2 = point(x=35, y=45)
p3 = point(x=60, y=25)
p4 = point(x=85, y=45)

s1 = polygon([p1, p2, p3, p4], fill="red", stroke="none")
s2 = polygon([p4, p3, p2, p1], fill="orange", stroke="none")

c2 = circle(r=50, fill="yellow", stroke="none")
c3 = circle(r=70, fill="red", stroke="none")
c4= circle(r=90, fill="green", stroke="none")

q1 = c4+c3+c2 + (s1 | repeat(20, rotate(20))) + (s2 | repeat(20, rotate(20)) | scale(0.85))

show(q1)
```
