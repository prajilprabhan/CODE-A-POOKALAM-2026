#Inner Pookalam Design


```python
p1 = point(x=10, y=25)
p2 = point(x=35, y=45)
p3 = point(x=60, y=25)
p4 = point(x=85, y=45)

s1 = polygon([p1, p2, p3, p4], fill="red", stroke="none")
s2 = polygon([p4, p3, p2, p1], fill="orange", stroke="none")

c2 = circle(r=50, fill="yellow", stroke="none")
c3 = circle(r=70, fill="red", stroke="none")
c4= circle(r=90, fill="green", stroke="none") | scale(0.88)

q1 = c4+c3+c2 + (s1 | repeat(20, rotate(20))) + (s2 | repeat(20, rotate(20)) | scale(0.85))

show(q1)
```

#Outer layer 1

```python
c=circle(x=50, y=0, r=100,fill="red")| repeat(18, rotate(angle=20))
r=circle(x=50, y=0, r=100) | repeat(18, rotate(angle=20))
c5=circle(x=50, y=0, r=70,stroke="white",fill="white")| repeat(18, rotate(angle=20))
c6=circle(x=0, y=0, r=140,stroke="green",stroke_width=5)
c2=circle(x=50, y=0, r=70,stroke="violet",stroke_width=3)| repeat(18, rotate(angle=20))

c3=circle(x=140,r=5,fill="yellow") | repeat(18, rotate(angle=20))
c4=circle(x=131,y=23,r=5,fill="green") | repeat(18, rotate(angle=20))
q2=c+r+c5+c2+c6+c3+c4 |scale(0.88)
show(q2)
```

```python
p1=point(x=0,y=5)
p2=point(x=10,y=5)
p4=point(x=10,y=10)
p3=point(x=5,y=40)
s1=polygon([p1,p2,p4,p3],fill="brown",stroke="none")
s2=polygon([p1,p2,p3],fill="brown",stroke="none") | translate(x=-15)|scale(0.8)
s3=polygon([p1,p2,p3],fill="brown",stroke="none") |translate(x=17)|scale(0.8)
r=s1+s2+s3
q3=r|translate(x=-4,y=-15)
show(q3)
```

#Outer layer

```python
p1 = point(x=70, y=10)
p2 = point(x=70, y=-10)

p3 = point(x=100, y=20)
p4 = point(x=100, y=-20)

p5 = point(x=140, y=0)

s1 = polygon([p1, p3, p5, p4, p2], fill="#ab0921",stroke="none")
s2 = polygon([p1, p3, p5, p4, p2],fill="red", stroke="none")
s3 = polygon([p1, p3, p5, p4, p2],    fill="green",    stroke="none")

r1 = s1
r2 = s2 | scale(0.92)
r3 = s3 | scale(0.84)

r4 = r1 + r2 + r3 | repeat(18, rotate(20))

c1 = circle(r=140, fill="yellow",stroke="none")
c2=circle(r=120,fill="white",stroke="yellow")
c3=circle(r=130,fill="green")
c4=circle(x=135,r=5,fill="#d6203b",stroke="#d6203b") | repeat(40,rotate(20))
c5 = rectangle (w=144,h=144,fill="white", stroke="white")|rotate(0) | repeat (15,rotate(50))



q4=c1+c3+c2+r4+c4+c5


show(q4)
```
